```python
from pypdf import PdfReader
from pdf2image import convert_from_path
import pytesseract
import os

def extract_pdf_text(pdf_path, dpi=300):
    """
    Extracts text from a PDF.
    1. First tries PyPDF (for digital PDFs).
    2. If no text, falls back to OCR (for scanned PDFs).
    """

    text = ""

    # 1️⃣ Try extracting with PyPDF (digital PDFs)
    try:
        reader = PdfReader(pdf_path)
        for page in reader.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
    except Exception as e:
        print(f"PyPDF failed: {e}")

    # 2️⃣ If still no text, fallback to OCR
    if not text.strip():
        print("No digital text found, using OCR...")
        try:
            pages = convert_from_path(pdf_path, dpi=dpi)
            for page in pages:
                text += pytesseract.image_to_string(page) + "\n"
        except Exception as e:
            print(f"OCR failed: {e}")

    return text.strip()
```

```python
pdf_file = "experian-sample-report.pdf"
text = extract_pdf_text(pdf_file)

print("Extracted text:\n")
print(text[:1000])  # print first 1000 chars
```
