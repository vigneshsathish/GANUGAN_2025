# Gradients

```python
x = Variable(torch.ones(2), requires_grad = True)
x
```

```python
print(x.size())
print(x.dtype)
```

```python
y = 5 *( x +1) **2
```

```python
o = (1/2) *  (torch.sum(y))
o
```

```python
print(o.size())
print(o.dtype)
```

```python
# calculate gradient with respect to x, since x 
o.backward()
```

```python
print(x.grad)
```
